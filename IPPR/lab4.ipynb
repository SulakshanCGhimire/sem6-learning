## Lab 4: Visualizing the effects of different smoothing (mean, median, guassian filter on images)

```python
import cv2
import numpy as numpy
import matplotlib.pyplot as plt
```

```python
# Step 1: Read image directly by path (should be in same folder as ipynb)
image_path = "IPPR_IMG.jpg"   
bgr_image = cv2.imread(image_path)

# Step 2: Verify it loaded
if bgr_image is None:
    print("Image not loaded! Check the filename or path.")
else:
    print("Image loaded! Shape:", bgr_image.shape)

    # Step 3: Convert & Display 
    rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
    plt.imshow(rgb_image)
    plt.axis('off')
    plt.title("Original Image")
    plt.show()
```

```python
# Mean Filter Visualization
# Step 4: Apply Mean Filter

mean_filtered = cv2.blur(rgb_image, ksize=(5, 5))

# Step 5: Display Mean Filtered Image
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(mean_filtered)
plt.title("Mean Filtered Image")
plt.axis('off')
plt.show()
```

```python
# Median Filter Visualization
# Step 6: Apply Median Filter
median_filtered = cv2.medianBlur(rgb_image, ksize=5)

# Step 7: Display Median Filtered Image
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(median_filtered)
plt.title("Median Filtered Image")
plt.axis('off')
plt.show()
```

```python
# Gaussian Filter Visualization
# Step 8: Apply Gaussian Filter
gaussian_filtered = cv2.GaussianBlur(rgb_image, ksize=(5, 5), sigmaX=0)

# Step 9: Display Gaussian Filtered Image
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(gaussian_filtered)
plt.title("Gaussian Filtered Image")
plt.axis('off')
```
