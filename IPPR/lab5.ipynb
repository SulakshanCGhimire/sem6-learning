## Lab 5: Implementation of Histogram Equalization in image using Python

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Function to display images using matplotlib
def display (title, img, cmap = 'gray'):
  plt.figure(figsize = (6,6))
  plt.imshow(img, cmap = cmap)
  plt.title(title)
  plt.axis('off')
  plt.show()

# Load the baseline image in Grayscale mode directly
image_path = "IPPR_IMG.jpg"
gray_img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if gray_img is None:
    print(f"Error: Could not load image from '{image_path}'. Check your file path!")
else:
    print("Baseline Grayscale environment initialized successfully!")

    # Show grayscale image
    display("Grayscale Image",gray_img)
```

```python
# Equalize the histogram of the grayscale image
equalized_img = cv2.equalizeHist(gray_img)

# Display the equalized image 
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap='gray')
plt.title("Gray Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(equalized_img, cmap='gray')
plt.title("Equalized Image")
plt.axis('off')
```

```python
# Displaying the histogram of the original and equalized images

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.hist(gray_img.ravel(), bins=256, color="#1f9bb43e", alpha=0.7)
plt.title("Histogram of Gray Image")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")
plt.subplot(1, 2, 2)
plt.hist(equalized_img.ravel(), bins=256, color="#82ff0ecc", alpha=0.7)
plt.title("Histogram of Equalized Image")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")
```
