# LAB 3: ALU Operations in Image Enhancement

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Function to display images using matplotlib
def display (title, img, cmap = 'gray'):
  plt.figure(figsize = (5,5))
  plt.imshow(img, cmap = cmap)
  plt.title(title)
  plt.axis('off')
  plt.show()
img1_raw = cv2.imread("IPPR_IMG1.jpg", cv2.IMREAD_GRAYSCALE)
img2_raw = cv2.imread("IPPR_IMG2.jpg", cv2.IMREAD_GRAYSCALE)

if img1_raw is None or img2_raw is None:
    print("Error: One or both images failed to load. Check file names and paths!")
else:
    # Match dimensions: Resize Image 2 to match Image 1's dimensions exactly
    height = 256
    width = 256
    img1 = cv2.resize(img1_raw, (width, height))
    img2 = cv2.resize(img2_raw, (width, height))
    
    print(f"Both images loaded and aligned to dimensions: {width}x{height}")
```

```python
# Perform ALU operations with the correctly loaded and resized images
add_img = cv2.add(img1, img2)
sub_img = cv2.subtract(img1, img2)
mul_img = cv2.multiply(img1, img2)
div_img = cv2.divide(img1, img2)

# Prepare titles and images for display
titles = ['Image 1', 'Image 2', 'Add', 'Subtract', 'Multiply', 'Divide']
images = [img1, img2, add_img, sub_img, mul_img, div_img]

# Display all images in a subplot grid
plt.figure(figsize = (15, 10))
for i in range(len(images)):
  plt.subplot(2, 3, i + 1) # 2 rows, 3 columns
  plt.imshow(images[i])
  plt.title(titles[i])
  plt.axis('off')
plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```
