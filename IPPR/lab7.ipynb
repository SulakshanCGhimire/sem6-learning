## Digital Image Restoration: Noise Reduction and Filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read image directly by path
image_path = "IPPR_IMG.jpg"   
bgr_image = cv2.imread(image_path)

# Verify it loaded
if bgr_image is None:
    print("Image not loaded! Check the filename or path.")
else:
    print("Image loaded! Shape:", bgr_image.shape)

    # Convert & Display 
    rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
    plt.imshow(rgb_image)
    plt.axis('off')
    plt.title("Original Image")
    plt.show()
    
    nl_means_denoised = cv2.fastNlMeansDenoisingColored(bgr_image, None, 10, 10, 7, 21)
    median_denoised = cv2.medianBlur(bgr_image, 3)
    bilateral_denoised = cv2.bilateralFilter(bgr_image, 9, 75, 75)
```

```python
titles = ["Original Image", "Non-Local Means Blur", "Median Blur", "Bilateral Denoising Blur"]
images = [rgb_image,
          cv2.cvtColor(nl_means_denoised, cv2.COLOR_BGR2RGB),
          cv2.cvtColor(median_denoised, cv2.COLOR_BGR2RGB), 
          cv2.cvtColor(bilateral_denoised, cv2.COLOR_BGR2RGB)]
plt.Figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')
    
plt.tight_layout()
plt.show()
```
